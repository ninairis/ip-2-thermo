```python
#met curvefit de waarde van n en I0 bepalen

q = -1.602*10**-19
k_B = 1.380649*10**-23
T = 293

def boltzmann_cf(U_diode,I0,n):
    exponent = (-q*U_diode)/(n*k_B*T)
    return I0*(np.exp(exponent)-1)

popt, pcov = curve_fit(boltzmann_cf, U_diode, I_diode, p0=[100^-9,2])
perr = np.sqrt(np.diag(pcov))

I0 = popt[0]
n = popt[1]

print(f'De gefitte waarde van I0 is {popt[0]} +/- {perr[0]}')
print(f'De gefitte waarde van n is {popt[1]} +/- {perr[1]}')
```
